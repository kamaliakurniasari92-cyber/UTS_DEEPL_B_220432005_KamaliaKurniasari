```python
# 202432005_KamaliaKurniasari

# Library untuk pengolahan data
import numpy as np
import pandas as pd
import re
import os
from glob import glob
from collections import Counter

# Library untuk visualisasi
import matplotlib.pyplot as plt

# Library untuk Deep Learning menggunakan PyTorch
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Library untuk pembagian dataset
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

Pada cell pertama, dilakukan proses import library yang dibutuhkan untuk menjalankan proses pengolahan data dan membangun model Deep Learning menggunakan algoritma Long Short-Term Memory (LSTM). Library NumPy dan Pandas digunakan untuk membantu proses manipulasi serta pengolahan data. Selanjutnya, Matplotlib digunakan untuk menampilkan visualisasi dalam bentuk grafik agar hasil analisis dapat diamati dengan lebih mudah.

Library PyTorch digunakan sebagai framework utama dalam membangun dan melatih model LSTM. Pada bagian ini, torch.nn digunakan untuk menyusun komponen jaringan saraf, sedangkan torch.optim digunakan untuk mengatur proses optimasi parameter model selama pelatihan. Selain itu, TensorDataset dan DataLoader digunakan untuk mengelola data dalam bentuk tensor dan membaginya ke dalam beberapa batch ketika proses training berlangsung.

Beberapa library tambahan seperti re, Counter, dan train_test_split juga digunakan untuk mendukung proses preprocessing teks, penyusunan vocabulary, serta pembagian dataset menjadi data training dan testing.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Dataset diunduh menggunakan KaggleHub. Jika unduhan meminta autentikasi atau gagal karena kebijakan akses, unggah Reviews.csv secara manual ke Colab dan ubah variabel csv_path pada cell berikutnya sesuai lokasi file. Notebook memuat hanya kolom Score dan Text untuk menghemat memori.</p>
</div>

```python
# 202432005_KamaliaKurniasari

# 1. Mengunduh dataset dari Kaggle
%pip install -q kagglehub

import kagglehub

dataset_path = kagglehub.dataset_download("snap/amazon-fine-food-reviews")

# Mencari file Reviews.csv di dalam folder dataset
csv_files = glob(os.path.join(dataset_path, "**", "Reviews.csv"), recursive=True)

if len(csv_files) == 0:
    raise FileNotFoundError("File Reviews.csv tidak ditemukan dalam folder dataset.")

csv_path = csv_files[0]

# Membaca kolom yang dibutuhkan
df = pd.read_csv(csv_path, usecols=["Text", "Score"])

print("Dataset berhasil dimuat!")
print("Lokasi dataset:", csv_path)
print("Jumlah data awal:", len(df))


# 2. Membersihkan data
df = df.dropna(subset=["Text", "Score"])
df = df.drop_duplicates(subset=["Text"])

# Mengubah rating menjadi label sentimen
# Rating 1-2 = Negatif (0)
# Rating 4-5 = Positif (1)
# Rating 3 = Netral (dihapus)

df = df[df["Score"] != 3].copy()

df["Sentiment"] = df["Score"].apply(
    lambda score: 0 if score <= 2 else 1
)

# Mengambil maksimal 15.000 data untuk setiap kelas
# Agar jumlah data lebih seimbang dan proses pelatihan lebih ringan

df_negative = df[df["Sentiment"] == 0].sample(
    n=min(15000, len(df[df["Sentiment"] == 0])),
    random_state=42
)

df_positive = df[df["Sentiment"] == 1].sample(
    n=min(15000, len(df[df["Sentiment"] == 1])),
    random_state=42
)

df = pd.concat([df_negative, df_positive], ignore_index=True)
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("\nJumlah data setelah pembersihan:", len(df))
print("\nDistribusi sentimen:")
print(df["Sentiment"].value_counts())


# 3. Case folding dan tokenisasi
def preprocess_text(text):
    # Mengubah seluruh teks menjadi huruf kecil
    text = str(text).lower()

    # Menghapus URL
    text = re.sub(r"http\S+|www\S+", "", text)

    # Menghapus angka dan tanda baca
    text = re.sub(r"[^a-z\s]", " ", text)

    # Menghapus spasi berlebih
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenisasi berdasarkan spasi
    tokens = text.split()

    return tokens


df["Tokens"] = df["Text"].apply(preprocess_text)

print("\nContoh hasil preprocessing:")
print("Teks asli :", df["Text"].iloc[0])
print("Token     :", df["Tokens"].iloc[0])


# 4. Membagi dataset menjadi data training dan testing
X_text = df["Tokens"].tolist()
y = df["Sentiment"].values

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("\nJumlah data training:", len(X_train_text))
print("Jumlah data testing :", len(X_test_text))


# 5. Membuat vocabulary mapping dari data training
MAX_VOCAB_SIZE = 20000
MAX_SEQUENCE_LENGTH = 120

# Menghitung frekuensi setiap kata pada data training
word_counts = Counter(
    word
    for sentence in X_train_text
    for word in sentence
)

# Menyediakan indeks khusus untuk padding dan kata tidak dikenal
word2idx = {
    "<PAD>": 0,
    "<UNK>": 1
}

# Memasukkan kata berdasarkan frekuensi kemunculannya
for word, count in word_counts.most_common(MAX_VOCAB_SIZE - 2):
    if count >= 2:
        word2idx[word] = len(word2idx)

print("\nJumlah vocabulary:", len(word2idx))


# 6. Mengubah token menjadi angka dan melakukan padding
def encode_and_pad(tokens, word2idx, max_length):
    # Mengubah setiap kata menjadi indeks vocabulary
    encoded = [
        word2idx.get(word, word2idx["<UNK>"])
        for word in tokens
    ]

    # Memotong teks jika melebihi panjang maksimum
    encoded = encoded[:max_length]

    # Menambahkan padding jika panjang teks kurang dari maksimum
    encoded += [word2idx["<PAD>"]] * (max_length - len(encoded))

    return encoded


X_train = np.array([
    encode_and_pad(text, word2idx, MAX_SEQUENCE_LENGTH)
    for text in X_train_text
])

X_test = np.array([
    encode_and_pad(text, word2idx, MAX_SEQUENCE_LENGTH)
    for text in X_test_text
])


# 7. Konversi data menjadi PyTorch Tensor
X_train = torch.tensor(X_train, dtype=torch.long)
X_test = torch.tensor(X_test, dtype=torch.long)

y_train = torch.tensor(y_train, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)

# Menampilkan ukuran tensor
print("\nUkuran Tensor:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

print("\nPreprocessing dataset berhasil diselesaikan!")
```

Pada cell kedua, dilakukan proses pemuatan dataset dan preprocessing terhadap data ulasan produk makanan dari Amazon Fine Food Reviews. Dataset diambil dari Kaggle menggunakan library KaggleHub, kemudian file Reviews.csv dibaca menggunakan Pandas dengan mengambil kolom Text dan Score yang diperlukan dalam proses klasifikasi sentimen.

Tahap selanjutnya adalah membersihkan data dengan menghapus baris yang memiliki nilai kosong dan menghilangkan data ulasan yang duplikat. Setelah itu, kolom Score digunakan untuk membentuk label sentimen. Ulasan dengan rating 1 dan 2 dikategorikan sebagai sentimen negatif dengan label 0, sedangkan rating 4 dan 5 dikategorikan sebagai sentimen positif dengan label 1. Ulasan dengan rating 3 tidak digunakan karena termasuk kategori netral.

Untuk menjaga keseimbangan jumlah data, dilakukan pengambilan maksimal 15.000 ulasan dari masing-masing kategori sentimen. Data kemudian diacak agar urutannya tidak berdasarkan kategori tertentu.

Setelah proses pembersihan selesai, dilakukan preprocessing teks melalui beberapa tahapan. Pertama, case folding untuk mengubah seluruh huruf menjadi huruf kecil. Kemudian, URL, angka, dan tanda baca dihapus agar teks lebih seragam. Selanjutnya, dilakukan tokenisasi dengan memecah setiap ulasan menjadi kumpulan kata berdasarkan spasi.

Data yang sudah ditokenisasi kemudian dibagi menjadi data training dan testing dengan perbandingan 80:20. Pembagian dilakukan menggunakan train_test_split dengan metode stratifikasi agar proporsi kelas positif dan negatif tetap terjaga pada kedua kelompok data.

Berikutnya, dibuat vocabulary mapping berdasarkan kata-kata yang terdapat pada data training. Setiap kata diberikan indeks angka agar dapat diproses oleh model LSTM. Indeks khusus juga disediakan untuk padding dan kata yang tidak ditemukan dalam vocabulary, yaitu <PAD> dan <UNK>.

Tahap berikutnya adalah mengubah setiap token menjadi urutan angka berdasarkan indeks vocabulary. Panjang setiap urutan diseragamkan menjadi 120 token menggunakan padding. Jika jumlah token melebihi batas tersebut, sebagian token akan dipotong, sedangkan jika jumlahnya kurang, akan ditambahkan nilai padding.

Terakhir, seluruh data input dan label dikonversi ke dalam format PyTorch Tensor. Data input menggunakan tipe torch.long karena berisi indeks kata, sedangkan label menggunakan torch.float32 agar dapat digunakan dalam proses perhitungan model klasifikasi. Hasil akhirnya berupa X_train, y_train, X_test, dan y_test yang akan digunakan pada tahap pembangunan dan pelatihan model LSTM.

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Arsitektur model terdiri atas tiga bagian. Pertama, Embedding mengubah setiap indeks kata menjadi vektor berdimensi tertentu. Kedua, LSTM mempelajari pola urutan kata dan hubungan konteks dalam ulasan. Ketiga, Linear menghasilkan satu nilai logit yang kemudian diubah menjadi probabilitas dengan sigmoid.

  Padding tidak dipakai sebagai representasi kata karena indeks 0 ditetapkan sebagai padding_idx. Pada proses forward, keluaran LSTM yang diambil adalah keluaran pada posisi token terakhir yang valid berdasarkan panjang kalimat, bukan posisi padding terakhir.</p>
</div>

```python
# 202432005_KamaliaKurniasari

class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim=100, hidden_dim=64, dropout=0.2):
        super(SequenceModel, self).__init__()

        # Layer embedding untuk mengubah indeks kata menjadi vektor
        self.embedding = nn.Embedding(
            num_embeddings=vocab_size,
            embedding_dim=embedding_dim,
            padding_idx=0
        )

        # Layer LSTM untuk mempelajari pola urutan kata
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_dim,
            batch_first=True
        )

        # Dropout untuk mengurangi risiko overfitting
        self.dropout = nn.Dropout(dropout)

        # Fully connected layer untuk menghasilkan prediksi
        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        # Mengubah indeks kata menjadi embedding
        embedded = self.embedding(x)

        # Memproses urutan kata menggunakan LSTM
        lstm_out, (hidden, cell) = self.lstm(embedded)

        # Mengambil hidden state terakhir
        last_hidden = hidden[-1]

        # Menerapkan dropout
        dropped = self.dropout(last_hidden)

        # Menghasilkan nilai prediksi
        output = self.fc(dropped)

        # Menghilangkan dimensi terakhir agar sesuai dengan label
        return output.squeeze(1)


# Inisialisasi model
vocab_size = len(word2idx)

model = SequenceModel(
    vocab_size=vocab_size,
    embedding_dim=100,
    hidden_dim=64,
    dropout=0.2
)

print(model)
```

Pada cell ketiga, dilakukan pembangunan arsitektur model Deep Learning menggunakan algoritma Long Short-Term Memory (LSTM). Model dibuat dengan mendefinisikan class SequenceModel yang merupakan turunan dari nn.Module pada PyTorch. Arsitektur ini digunakan untuk mempelajari hubungan antarurutan kata dalam ulasan produk sehingga model dapat melakukan klasifikasi sentimen positif dan negatif.

Tahap pertama adalah membuat layer Embedding menggunakan nn.Embedding. Layer ini berfungsi mengubah setiap indeks kata yang sebelumnya dihasilkan pada proses preprocessing menjadi representasi vektor numerik. Pada implementasi ini, ukuran embedding ditetapkan sebesar 100 dimensi. Parameter padding_idx=0 digunakan agar indeks padding tidak ikut diperbarui selama proses pelatihan.

Selanjutnya, dibuat layer LSTM menggunakan nn.LSTM dengan jumlah hidden state sebanyak 64. Parameter batch_first=True digunakan agar format input mengikuti urutan batch, sequence, dan embedding. Melalui layer ini, model mempelajari pola hubungan antar kata dalam sebuah ulasan dengan memanfaatkan mekanisme memori LSTM.

Setelah proses LSTM, digunakan layer Dropout sebesar 0,2 untuk membantu mengurangi risiko overfitting. Artinya, selama proses training, sebagian nilai neuron akan dinonaktifkan secara acak sesuai dengan probabilitas yang ditentukan.

Berikutnya, digunakan Fully Connected Layer (nn.Linear) untuk mengubah representasi hasil LSTM menjadi satu nilai output. Nilai tersebut merupakan skor prediksi yang nantinya digunakan untuk menentukan sentimen positif atau negatif.

Pada fungsi forward(), data input terlebih dahulu melewati layer Embedding, kemudian diproses oleh LSTM. Dari hasil pemrosesan tersebut, diambil hidden state terakhir melalui hidden[-1] sebagai representasi akhir dari urutan kata. Hasilnya kemudian melewati Dropout dan Fully Connected Layer.

Fungsi squeeze(1) digunakan untuk menghilangkan dimensi tambahan pada output sehingga bentuk hasil prediksi sesuai dengan bentuk label yang digunakan dalam proses pelatihan.

Terakhir, model diinisialisasi menggunakan jumlah vocabulary yang telah dibuat pada Cell 2. Dengan demikian, model LSTM sudah siap digunakan pada tahapan berikutnya, yaitu proses konfigurasi dan pelatihan model.

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Pelatihan dilakukan dengan fungsi loss BCEWithLogitsLoss karena target berupa dua kelas dan model menghasilkan satu logit. Optimizer Adam digunakan untuk memperbarui bobot model berdasarkan gradien. Data validation dipakai untuk memantau kinerja pada data yang tidak digunakan untuk memperbarui bobot.

  Nilai hyperparameter awal: embedding 100 dimensi, hidden size 64, batch size 128, learning rate 0,001, dan maksimum 5 epoch. Jumlah epoch dapat ditambah setelah memeriksa waktu pelatihan dan pola loss.</p>
</div>

```python
# 202432005_KamaliaKurniasari

# 1. Inisialisasi hyperparameter
epochs = 5
learning_rate = 0.001
batch_size = 64
hidden_size = 64

# Menentukan perangkat yang digunakan (GPU jika tersedia)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Perangkat yang digunakan:", device)


# 2. Menyiapkan dataset dan DataLoader
train_dataset = TensorDataset(X_train, y_train)

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

# Memindahkan model ke perangkat yang digunakan
model = model.to(device)


# 3. Inisialisasi Loss Function dan Optimizer
criterion = nn.BCEWithLogitsLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=learning_rate
)


# 4. Menyimpan riwayat training loss
train_losses = []


# 5. Menjalankan proses training loop
for epoch in range(epochs):

    model.train()
    running_loss = 0.0

    for inputs, labels in train_loader:

        # Memindahkan data ke GPU atau CPU
        inputs = inputs.to(device)
        labels = labels.to(device)

        # Menghapus gradient dari iterasi sebelumnya
        optimizer.zero_grad()

        # Forward pass
        outputs = model(inputs)

        # Menghitung nilai loss
        loss = criterion(outputs, labels)

        # Backward pass
        loss.backward()

        # Membatasi gradient agar training lebih stabil
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        # Memperbarui parameter model
        optimizer.step()

        # Mengakumulasi nilai loss
        running_loss += loss.item() * inputs.size(0)

    # Menghitung rata-rata loss setiap epoch
    epoch_loss = running_loss / len(train_dataset)

    # Menyimpan loss ke dalam list
    train_losses.append(epoch_loss)

    print(
        f"Epoch [{epoch + 1}/{epochs}], "
        f"Training Loss: {epoch_loss:.4f}"
    )

print("\nProses training model LSTM selesai!")
```

Pada cell keempat, dilakukan proses konfigurasi parameter dan pelatihan model LSTM menggunakan data ulasan produk makanan Amazon yang telah melalui tahap preprocessing. Proses pelatihan bertujuan agar model dapat mempelajari pola hubungan antar kata dalam ulasan sehingga mampu membedakan sentimen positif dan negatif.

Tahap pertama adalah menentukan hyperparameter yang digunakan selama proses pelatihan. Parameter epochs ditetapkan sebanyak 5, yang berarti seluruh data training akan diproses sebanyak lima kali. Selanjutnya, learning_rate ditetapkan sebesar 0,001 untuk mengatur besarnya langkah pembaruan bobot model. batch_size ditetapkan sebesar 64, sehingga data training diproses dalam kelompok yang masing-masing berisi maksimal 64 data. Parameter hidden_size sebesar 64 menunjukkan jumlah unit pada hidden layer LSTM yang telah ditentukan pada Cell 3.

Selanjutnya, program menentukan perangkat komputasi yang digunakan melalui torch.device. Jika GPU tersedia, proses pelatihan akan dijalankan menggunakan CUDA. Namun, apabila GPU tidak tersedia, program akan menggunakan CPU.

Tahap berikutnya adalah menyiapkan data training menggunakan TensorDataset dan DataLoader. TensorDataset menggabungkan data input dan label menjadi satu dataset, sedangkan DataLoader membagi dataset tersebut menjadi beberapa batch. Parameter shuffle=True digunakan untuk mengacak urutan data pada setiap epoch agar model tidak mempelajari data berdasarkan urutan tertentu.

Setelah itu, model dipindahkan ke perangkat komputasi yang telah ditentukan. Fungsi loss yang digunakan adalah BCEWithLogitsLoss, yaitu fungsi loss yang sesuai untuk klasifikasi biner dengan satu nilai output berupa logit. Fungsi ini menggabungkan perhitungan sigmoid dan binary cross-entropy sehingga proses perhitungan loss dapat dilakukan secara lebih stabil.

Optimizer yang digunakan adalah Adam dengan learning rate sebesar 0,001. Optimizer ini bertugas memperbarui bobot dan parameter model berdasarkan gradient yang dihasilkan selama proses pelatihan.

Selanjutnya, dibuat list train_losses untuk menyimpan nilai rata-rata loss pada setiap epoch. List tersebut nantinya digunakan untuk membuat grafik riwayat training loss sehingga perkembangan proses pembelajaran model dapat diamati.

Proses training dilakukan melalui perulangan sebanyak jumlah epoch yang telah ditentukan. Pada setiap epoch, model diatur ke mode training menggunakan model.train(). Kemudian, setiap batch data dipindahkan ke perangkat komputasi dan diproses melalui tahapan forward pass untuk menghasilkan prediksi.

Hasil prediksi dibandingkan dengan label sebenarnya menggunakan fungsi loss. Setelah nilai loss diperoleh, dilakukan backward pass melalui loss.backward() untuk menghitung gradient dari setiap parameter model. Fungsi clip_grad_norm_ digunakan untuk membatasi nilai gradient maksimal sebesar 1,0 agar proses pembaruan parameter lebih stabil.

Berikutnya, optimizer.step() digunakan untuk memperbarui bobot model berdasarkan hasil perhitungan gradient. Nilai loss dari setiap batch kemudian diakumulasikan dan dihitung rata-ratanya pada akhir setiap epoch. Hasil rata-rata tersebut disimpan ke dalam train_losses dan ditampilkan agar perkembangan proses pelatihan dapat dipantau.

Setelah seluruh epoch selesai dijalankan, model telah melalui proses pembelajaran menggunakan data training dan siap memasuki tahap evaluasi.

```python
# 2. Menghasilkan Prediksi pada Data Testing
model.eval()

test_outputs = []

with torch.no_grad():
    for start in range(0, len(X_test), 64):
        inputs = X_test[start:start + 64].to(device)

        outputs = model(inputs)

        probabilities = torch.sigmoid(outputs)

        test_outputs.append(probabilities.cpu())

# Menggabungkan seluruh hasil prediksi
predicted_probs = torch.cat(test_outputs).numpy()

# Mengubah probabilitas menjadi kelas sentimen
predicted_labels = (predicted_probs >= 0.5).astype(int)

# Mengambil label aktual dari data testing
actual_labels = y_test.numpy().astype(int)
```
